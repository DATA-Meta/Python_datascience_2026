```python
# importing pandas library
import pandas as pd 
```

```python
# create a sample data set
data = {'Age': [66, 32,27, 40, 51, 22], 
        'Owned_House': ['yes', 'yes', 'no', 'no', 'yes', 'yes'],
        'Salary': [54000, 321250, 29120, 46000, 61000, 52000],
        'Martial_status': ['widowed', 'married', 'married', 'single', 'divorced', 'single'],
        'Distance to the nearest town(km)': [5.4, 10.2, 6.7, 5.7, 15, 26.5]}
        
# creating a dataframe from the data
df = pd.DataFrame(data)

# display the dataframe
df
```

```python
# checking the data types of the columns
data_types = df.dtypes

data_types
```

```python
# checking the data types of the columns
Age_data_type = df['Age'].dtype
print(f"The data type of the 'Age' column is: {Age_data_type}")
```

```python
# retrieve the particular data types by using .select_dtypes() method 
select_attributes = df.select_dtypes(include=['object', 'float64'])
select_attributes
```

 **Here above in the output the data shows 'Nominal data ' which have attributes namely Owned_House , Martial_status. Continous data in the output for attribute for 'Distance to the nearest town(km)** 

```python
# Example 1: Finding missing data with Boolean results
import pandas as pd
#Importing numpy library
import numpy as np

data = {'Age': [np.nan, 32, 27, 40, 51, 22],
        'Owned_House': ['yes', 'yes', 'no',np.nan, 'yes', 'yes'],
        'Salary': [54000, 321250, 29120, 46000, 61000, 52000],
        'Martial_status': ['widowed', np.nan,np.nan, 'single', 'divorced', 'single'],
        'Distance to the nearest town(km)': [5.4, 10.2, 6.7, np.nan, 15, 26.5]}


df = pd.DataFrame(data)

df

#checking null values gives TRUE 
print(df.isnull())

#checking non missing value notnull()
print(df.notnull())
```
