# Using simple linear regression 

```python
from sklearn.linear_model import LinearRegression
```

```python
# loading dataset od iris flower
df = sns.load_dataset("tips")
df.head()
```

```python
# seaborn ploting using scatterplot 
sns.scatterplot(x = 'total_bill',y = 'tip', data =df)
```

```python
# Selecting features from df
x = df[['total_bill']]
y = df[['tip']]
```

```python
# create a regression line plot using seaborn
sns.regplot(x = 'total_bill',y = 'tip', data =df)
```

```python
x = df[['total_bill']]
y = df[['tip']]
```

```python
#load the model
model = LinearRegression()
# fit the model means train the model 
model.fit(x,y)
```

```python
# lets predict the tip for unknown total bill
model.predict([[50]])
```
