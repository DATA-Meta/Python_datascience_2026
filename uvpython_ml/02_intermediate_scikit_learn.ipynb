## Simple Linear Regression

```python
# importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
df = sns.load_dataset('tips') 
df.head()
# encode smoker column
df['smoker_num'] = df['smoker'].map({'Yes': 1, 'No': 0})
df.head()
```

```python
# creating x and y variables
x = df[['total_bill', 'size', 'smoker_num']]
y = df[['tip']]
```

```python
#load the model
model = LinearRegression()
#train the model
model.fit(x,y)

# predict the model for unknown total bill
print(model.predict([[35.527, 2, 1]]))
print(model.predict([[35.527,2,0]]))
```

```python
# plotting the regression line as well as the data points using both the features total_bill and size and smoker_num 
sns.lmplot(x = 'total_bill', y = 'tip', data = df, hue = 'size', palette = 'Set1')
```
